```python
import numpy as np
data = np.array([[5.1, 1.4, 0],
                 [4.9, 1.4, 0],
                 [4.7, 1.3, 0],
                 [4.6, 1.5, 0],
                 [5.0, 1.4, 0],
                 [7.0, 4.7, 1],
                 [6.4, 4.5, 1],
                 [6.9, 4.9, 1],
                 [5.5, 4.0, 1],
                 [6.5, 4.6, 1]])
X = data[:,[0,1]]
y = data[:,2]

w_ = np.zeros(X.shape[1])
b_ = np.float64(0.)
losses_ = []
n_iter = 2
eta = 0.01
rgen = np.random.RandomState(1)

for i in range(n_iter):
    # Чтобы не вычислять среднее значение суммы накопленных ошибок во всём наборе данных как в ALN
    # можно пошагово как в MCP обновляем параметры для каждого обучающего примера - например:
    # дельта(w) = скорость обучения * (yi - Zi)*xi
    # дельта(b) = скорость обучения * (yi - Zi)

    # Чтобы получить удовлетворительные результаты с помощью SGD, важно представлять
    # обучающие данные в случайном порядке в каждой эпохе
    r = rgen.permutation(len(y))
    X,y = X[r], y[r]
    losses = []

    for xi, target in zip(X, y):
        """Применение правила Adaline для обновления весов"""
        error = (target - np.dot(xi,w_) + b_)
        w_ += eta * 2.0 * xi * (error)
        b_ += eta * 2.0 * error
        loss = error**2
        losses.append(loss)
        print("ошибка", error, "весы", w_, "смещение", b_, "квадратичная ошибка", loss)

    avg_loss = np.mean(losses)
    print("средне квадратичная ошибка за эпоху", avg_loss)
```
