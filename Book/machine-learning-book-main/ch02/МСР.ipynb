```python
import numpy as np
data = np.array([[5.1, 1.4, 0],
                 [4.9, 1.4, 0],
                 [4.7, 1.3, 0],
                 [4.6, 1.5, 0],
                 [5.0, 1.4, 0],
                 [7.0, 4.7, 1],
                 [6.4, 4.5, 1],
                 [6.9, 4.9, 1],
                 [5.5, 4.0, 1],
                 [6.5, 4.6, 1]])
X = data[:,[0,1]]
y = data[:,2]

w_ = np.zeros(X.shape[1])
b_ = np.float64(0.)
errors_ = []
n_iter = 10
eta = 0.1

for _ in range(n_iter):
    errors = 0
    for xi, target in zip(X, y):
        s = np.dot(xi, w_) + b_
        update = eta * (target - np.where(s >= 0.0, 1, 0))
        w_ += update * xi
        b_ += update
        errors += int(update != 0.0)
        print("признаки",xi,"функция",s, "дельта",update,"весы",w_,"смещение",b_)
    errors_.append(errors)

errors_
```
